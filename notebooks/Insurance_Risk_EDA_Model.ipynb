# Insurance Claim Risk Classification - EDA & Modeling

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

## 2. Load Dataset

```python
df = pd.read_csv('../data/insurance_claims.csv')
df.head()
```

## 3. Exploratory Data Analysis (EDA)

```python
# Check for missing values
df.isnull().sum()
```

```python
# Distribution of Risk Labels
sns.countplot(x='Risk_Label', data=df)
plt.title('Distribution of High vs Low Risk Claims')
plt.show()
```

```python
# Claim Amount by Risk
plt.figure(figsize=(10, 6))
sns.boxplot(x='Risk_Label', y='Claim_Amount', data=df)
plt.title('Claim Amount vs Risk')
plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(12, 8))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

## 4. Data Preprocessing

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

# Encoding Categorical Variables
le = LabelEncoder()
for col in ['Gender', 'Policy_Type', 'Claim_Type', 'Risk_Label']:
    df[col] = le.fit_transform(df[col])

# Dropping ID
X = df.drop(['Claim_ID', 'Risk_Label'], axis=1)
y = df['Risk_Label']

# Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Splitting
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

## 5. Model Building

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

models = {
    'Logistic Regression': LogisticRegression(),
    'Decision Tree': DecisionTreeClassifier(),
    'Random Forest': RandomForestClassifier()
}

results = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    results[name] = acc
    print(f"{name} Accuracy: {acc:.4f}")
    print(classification_report(y_test, y_pred))
```

## 6. Model Evaluation

```python
# Compare Models
plt.bar(results.keys(), results.values())
plt.title('Model Accuracy Comparison')
plt.ylim(0, 1)
plt.show()
```
